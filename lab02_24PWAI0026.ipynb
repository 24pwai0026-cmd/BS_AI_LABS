```python
# Task 02: Simple Mini Calculator

# Ask user for two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
op = input("Enter operation (+, -, *, /, **, %): ")

if op not in ["+", "-", "*", "/", "**", "%"]:
    print("Invalid operator!")

# Check division by zero
elif op == "/" and num2 == 0:
    print("Cannot divide by zero!")

else:
    # Perform calculation
    if op == "+":
        result = num1 + num2
        print(f"{num1} + {num2} = {result}")

    elif op == "-":
        result = num1 - num2
        print(f"{num1} - {num2} = {result}")

    elif op == "*":
        result = num1 * num2
        print(f"{num1} * {num2} = {result}")

    elif op == "/":
        result = num1 / num2
        print(f"{num1} / {num2} = {result:.2f}")

    elif op == "**":
        result = num1 ** num2
        print(f"{num1} ** {num2} = {result:.2f}")

    elif op == "%":
        result = num1 % num2
        print(f"{num1} % {num2} = {result}")
```

```python
# Ask user for starting and ending numbers
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))
while start <= end:
    if start % 7 == 0:
        start += 1
        continue

    # Check conditions
    if start % 2 == 0 and start % 5 == 0:
        print(f"{start} : Even & Multiple of 5!!")

    elif start % 2 == 0:
        print(f"{start} : Even")

    elif start % 5 == 0:
        print(f"{start} : Multiple of 5!")

    else:
        print(start)

    start += 1
```

```python
# Ask user for password
password = input("Enter password: ")
if len(password) < 6:
    print("Too short!")
else:
    has_digit = False
    has_upper = False

    # Check characters in password
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True
    # Conditions
    if not has_digit:
        print("Add at least one number")

    elif not has_upper:
        print("Add at least one capital letter")

    else:
        print("Strong password – good choice!")
```
