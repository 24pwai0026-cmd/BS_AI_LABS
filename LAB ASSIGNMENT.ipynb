```python
# NumPy for AI - Mini Project Assignment

import numpy as np

# STEP 1: Create Dataset of 10 Students
# Columns:

data = np.array([
    [78, 85, 80, 90, 1],
    [45, 50, 40, 55, 0],
    [88, 92, 84, 91, 1],
    [60, 65, 58, 62, 1],
    [35, 40, 30, 45, 0],
    [95, 90, 93, 97, 1],
    [55, 48, 52, 50, 0],
    [72, 70, 75, 78, 1],
    [40, 42, 38, 35, 0],
    [82, 79, 85, 88, 1]
])

print("Dataset:")
print(data)

# STEP 2: Separate Features (X) and Labels (y)
X = data[:, 0:4]
y = data[:, 4]

print("Features (X):")
print(X)

print("Labels (y):")
print(y)

# STEP 3: Calculate Average, Maximum, Minimum Marks

average_marks = np.mean(X, axis=0)
maximum_marks = np.max(X, axis=0)
minimum_marks = np.min(X, axis=0)

subjects = ["Math", "Science", "English", "Computer"]

print("Average Marks:")
for i in range(4):
    print(subjects[i], "=", average_marks[i])

print("Maximum Marks:")
for i in range(4):
    print(subjects[i], "=", maximum_marks[i])

print("Minimum Marks:")
for i in range(4):
    print(subjects[i], "=", minimum_marks[i])

# STEP 4: Count Passed and Failed Students

passed_students = np.sum(y == 1)
failed_students = np.sum(y == 0)

print("Number of Passed Students =", passed_students)
print("Number of Failed Students =", failed_students)

# STEP 5: Students Scoring Above 80 and Below 50
above_80 = X[X > 80]
below_50 = X[X < 50]

print("Marks Above 80:")
print(above_80)

print("Marks Below 50:")
print(below_50)
# STEP 6: Normalize Dataset
# Formula:
# x_norm = (x - min) / (max - min)


X_norm = (X - np.min(X)) / (np.max(X) - np.min(X))

print("Normalized Dataset:")
print(X_norm)

# STEP 7: Simple Prediction Logic
# If average marks >= 60 → Pass (1)
# Else → Fail (0)


average_student_marks = np.mean(X, axis=1)

predictions = np.where(average_student_marks >= 60, 1, 0)

print("Predicted Results:")
print(predictions)

# STEP 8: Calculate Accuracy
accuracy = np.mean(predictions == y) * 100

print("Actual Results:")
print(y)

print("Accuracy =", accuracy, "%")
```
