```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =========================
# LINE GRAPH
# =========================

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [30, 32, 31, 35, 33]

plt.plot(days, temperature, marker='o')
plt.title("Temperature Changes Over 5 Days")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid(True)
plt.show()


# =========================
# BAR CHART
# =========================

students = ["Ali", "Sara", "Ahmed", "Ayesha"]
marks = [78, 85, 90, 72]
colors = ["red", "blue", "green", "orange"]

bars = plt.bar(students, marks, color=colors)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             str(bar.get_height()),
             ha='center')

plt.show()


# =========================
# HISTOGRAM
# =========================

scores = np.random.randint(0, 100, 500)

plt.hist(scores, bins=10)
plt.title("Exam Scores Histogram")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()


# =========================
# SCATTER PLOT
# =========================

hours = [1,2,3,4,5,6,7]
marks2 = [40,45,50,60,70,80,90]

plt.scatter(hours, marks2, color='red')
plt.title("Study Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()


# =========================
# SEABORN LINE PLOT
# =========================

x = [1,2,3,4,5]
y = [10,20,25,30,40]

sns.lineplot(x=x, y=y, marker='o')
plt.title("Seaborn Line Plot")
plt.show()


# =========================
# HEATMAP
# =========================

matrix = np.random.rand(5,5)

sns.heatmap(matrix, annot=True, cmap="coolwarm")
plt.title("5x5 Heatmap")
plt.show()


# =========================
# MATPLOTLIB VS SEABORN
# =========================

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.scatter(x, y)
plt.title("Matplotlib Scatter")
plt.show()

sns.scatterplot(x=x, y=y)
plt.title("Seaborn Scatter")
plt.show()

print("Seaborn is more visually appealing.")
print("Matplotlib gives more control.")


# =========================
# MINI DATASET
# =========================

names = ["Ali","Sara","Ahmed","Ayesha","Hamza"]
marks = [78,85,90,72,88]
attendance = [80,90,95,70,85]

plt.bar(names, marks)
plt.title("Marks")
plt.show()

plt.plot(names, attendance, marker='o')
plt.title("Attendance Trend")
plt.show()

data = np.array([marks, attendance])

sns.heatmap(data, annot=True, xticklabels=names,
            yticklabels=["Marks","Attendance"])
plt.title("Marks vs Attendance")
plt.show()
```
